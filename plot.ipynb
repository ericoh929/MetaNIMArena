```python
# 필요한 라이브러리 다시 불러오기
import matplotlib.pyplot as plt
import numpy as np

# 그래프 설정
fig, axes = plt.subplots(2, 2, figsize=(7, 5), sharey=True, dpi=300)
axes = axes.flatten()

# 상태 및 액션 정의
states = ["State 1 (Wrong Bias)", "State 2 (Wrong Bias)", "State 3 (Good Bias)", "State 4 (Good Bias)"]
actions = ["Action 1", "Action 2", "Action 3"]

# 데이터 정의
consistency_data = np.array([
    [15, 2.5, 82.5],   # State 1
    [2.5, 10, 87.5],  # State 2
    [85, 7.5, 7.5],  # State 3
    [2.5, 27.5, 70]  # State 4
])

debate_data = np.array([
    [10, 0, 90],   # State 1
    [0, 10, 90],   # State 2
    [90, 0, 10],   # State 3
    [0, 20, 80]   # State 4
])

# 색상 및 패턴 설정
custom_colors = ["#9467bd", "#1f77b4"]  # Self-Consistency, Multi-Agent Debate
bar_width = 0.35  # 막대 그래프 너비
x_labels = np.arange(len(actions))  # X축 위치
hatch_patterns = ["//", "\\\\"]  # 패턴 추가

# 그래프 생성
for i, state in enumerate(states):
    ax = axes[i]
    bars1 = ax.bar(x_labels - bar_width/2, consistency_data[i], bar_width, alpha=0.85,
                   color=custom_colors[0], hatch=hatch_patterns[0], edgecolor="black", linewidth=2)
    bars2 = ax.bar(x_labels + bar_width/2, debate_data[i], bar_width, alpha=0.85,
                   color=custom_colors[1], hatch=hatch_patterns[1], edgecolor="black", linewidth=2)

    # 각 바 위에 값 표시
    for bar in bars1:
        ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 2, f"{bar.get_height():.1f}",
                ha='center', fontsize=7, fontweight="bold")
    for bar in bars2:
        ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 2, f"{bar.get_height():.1f}",
                ha='center', fontsize=7, fontweight="bold")

    # X축 설정
    ax.set_xticks(x_labels)
    ax.set_xticklabels(actions)
    ax.set_ylabel("Action Frequency (%)")
    ax.set_ylim(0, 100)
    ax.set_title(state, fontsize=12, fontweight="bold")

    # 범례 설정: 왼쪽 위 그래프(첫 번째 그래프)만 표시
    if i == 0:
        ax.legend(["Self-Consistency", "Multi-Agent Debate"], fontsize=8, loc="upper left", frameon=False)

# 전체 제목 및 레이아웃 조정
plt.suptitle("Comparison of Action Distributions: Self-Consistency vs. Multi-Agent Debate",
             fontsize=10, fontweight="bold")
plt.tight_layout(rect=[0, 0.03, 1, 0.95])

# 그래프 표시
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Given data
data = {
    "a_0": [0.273, 0.291, 0.364],
    "a_2": [0.364, 0.309, 0.291],
    "a_4": [0.345, 0.291, 0.364],
    "a_6": [0.382, 0.309, 0.291],
    "a_8": [0.345, 0.364, 0.345],
    "a_10": [0.400, 0.309, 0.327],
}

# Calculate means and standard deviations
x_values = [int(k.split('_')[1]) for k in data.keys()]
y_values = [np.mean(v) for v in data.values()]
y_std = [np.std(v) for v in data.values()]

# Create figure
plt.figure(figsize=(8, 5), dpi=300)

# Use a modern style for better aesthetics
plt.style.use("ggplot")

# Plot with error bars
plt.errorbar(
    x_values, y_values, yerr=y_std, fmt='o-', linewidth=2.5, markersize=8,
    color='royalblue', ecolor='gray', capsize=6, capthick=2, elinewidth=2, label="Mean ± Std"
)

# Fill between for a smooth standard deviation representation
plt.fill_between(x_values, np.array(y_values) - np.array(y_std), np.array(y_values) + np.array(y_std),
                 color='gray', alpha=0.2)

# Styling
plt.xlabel("Variable Index", fontsize=14, fontweight='bold')
plt.ylabel("Mean Value", fontsize=14, fontweight='bold')
plt.title("Mean Values with Standard Deviation", fontsize=16, fontweight='bold', pad=15)
plt.xticks(fontsize=12)
plt.yticks(fontsize=12)
plt.grid(True, linestyle='--', linewidth=0.6, alpha=0.7)

# Improved annotation style
for x, y in zip(x_values, y_values):
    plt.annotate(f'{y:.3f}', (x, y), textcoords="offset points", xytext=(0, 8),
                 ha='center', fontsize=12, fontweight='bold', color='black')

# Save high-quality figure
plt.legend(fontsize=12, loc='upper right')
plt.tight_layout()
plt.show()
```

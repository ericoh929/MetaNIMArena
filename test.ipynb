````python
import re, json
matches_with_braces = re.search(r'\{.*?\}', text, re.DOTALL)

print(matches_with_braces)

````

````python
parsed_content_with_braces = matches_with_braces.group(0) if matches_with_braces else None

print(type(parsed_content_with_braces), parsed_content_with_braces)

parsed_content_with_braces = parsed_content_with_braces.replace('\xa0', '').strip()

print(type(parsed_content_with_braces), parsed_content_with_braces)
````
